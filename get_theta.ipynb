```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
filename = "picklehello"
```

```python
with open(filename, 'rb') as file:
    pickle_data = pickle.load(file)

print(pickle_data)
```

```python
for snap in range(200,201):
    with open(filename, 'rb') as file:
        pickle_data = pickle.load(file)
    
```

```python
snapshots = []
thetas = []

for snap_key, snap_data in pickle_data.items():

    snap_num = int(snap_key.replace("snap", ""))
    
    theta_val = snap_data["theta"]
    
    snapshots.append(snap_num)
    thetas.append(theta_val)

snapshots = np.array(snapshots)
thetas = np.array(thetas, dtype=float)

sort_idx = np.argsort(snapshots)
snapshots = snapshots[sort_idx]
thetas = thetas[sort_idx]

mask = ~np.isnan(thetas)
snapshots = snapshots[mask]
thetas = thetas[mask]

plt.figure(figsize=(8,5))
plt.plot(snapshots, thetas)
plt.xlabel("Snapshot")
plt.ylabel("Theta")
plt.title("Theta vs Snapshot")
# plt.grid(True)
plt.axvline(287, color='red', alpha=0.3)
plt.axvline(300, color='red', alpha=0.3)
plt.axvline(449, color='red', alpha=0.3)
plt.axvline(472, color='red', alpha=0.3)
plt.axhline(90, color='green')
plt.show()
```
