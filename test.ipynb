```python
import gizmo_analysis as gizmo
import utilities as ut

import numpy as np
 
import matplotlib.pyplot as plt

import os 

import pickle
```

```python
def theta_warps(snapshot):
    simulation_directory = '/home/eriley1/bighome/warp-data/m12b'

    part = gizmo.io.Read.read_snapshots('star', 'index', snapshot, simulation_directory, assign_hosts_rotation=True,
                                        assign_hosts=True)

    SpeciesProfile = ut.particle.SpeciesProfileClass(

    limits=[0, 20], width=1, log_scale=False, dimension_number=3)


    pro = SpeciesProfile.get_sum_profiles(part, 'star', 'mass')

    r = part['star'].prop('host.distance.total')
    R = part['star'].prop('host.distance.principal.cylindrical')[:,0]
    x = part['star'].prop('host.distance.principal')[:,0]
    y = part['star'].prop('host.distance.principal')[:,1]
    z = part['star'].prop('host.distance.principal')[:,2]
    mass = part['star'].prop('mass')

    vs=part['star'].prop('host.velocity.principal')
    rs=part['star'].prop('host.distance.principal')

    m = part['star'].prop('mass')

    L = np.cross(rs, vs) * m[:, None]
    Lmag = np.linalg.norm(L, axis=1)
    Lz_over_L = np.divide(L[:,2], Lmag, out=np.zeros_like(Lmag), where=Lmag>0)

    Lx=L[:,0]
    Ly=L[:,1]
    Lz=L[:,2]

    prof_lx = SpeciesProfile.get_statistics_profile(properties=Lx,distances=r) 
    prof_ly= SpeciesProfile.get_statistics_profile(properties=Ly,distances=r)
    prof_lz = SpeciesProfile.get_statistics_profile(properties=Lz,distances=r)

    mean_lx=prof_lx['average']
    mean_ly=prof_ly['average']
    mean_lz=prof_lz['average']

    dists=prof_lx["distance.mid"]

    prof_mass = SpeciesProfile.get_statistics_profile(properties=mass,distances=r) 
    mean_mass=prof_mass['average']

    L_profile_mag=np.sqrt(mean_lx**2+mean_ly**2+mean_lz**2)

    theta=np.arccos(np.divide(mean_lz,L_profile_mag, out=np.zeros_like(L_profile_mag), where=L_profile_mag>0))*(180/np.pi)
    # plt.plot(dists,theta,label='theta',color='black')
    # plt.xlabel('r [kpc]')
    # plt.ylabel(r'$\theta$ [degrees]')

    num = np.sum(theta * mean_mass)
    den = np.sum(mean_mass)

    theta_fin = num / den
    print(theta_fin, 'hello')
    print(dists)
    print(mean_mass)

    return theta_fin, dists, mean_mass
```

```python
for i in range(49,50):
    try:
        theta_fin,dists,mean_mass = theta_warps(i)
    
        temp_values = {}
        temp_values={"theta":theta_fin,
                 "dists":dists,
                 "mean_mass":mean_mass}

        values["snap"+str(i)] = temp_values
        # values.update("snap"+str(i):temp_values)
        print(temp_values)

    except Exception as e:
        continue
```
